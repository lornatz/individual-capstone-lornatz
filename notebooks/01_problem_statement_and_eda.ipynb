## Section 3: Initial EDA

### 3.1 Target Variable Analysis

```python
# Define target variable
TARGET = 'price'

# Basic statistics of target
print(f"Target Variable: {TARGET}")
print(f"\nBasic Statistics:")
print(df[TARGET].describe())

print(f"\nAdditional Statistics:")
print(f"Median: ${df[TARGET].median():,.2f}")
print(f"Mode: ${df[TARGET].mode().values[0]:,.2f}")
print(f"Std Dev: ${df[TARGET].std():,.2f}")
```

```python
# Plot distribution of target variable
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histogram
axes[0].hist(df[TARGET].dropna(), bins=50, edgecolor='black', alpha=0.7, color='steelblue')
axes[0].set_xlabel('Price ($)', fontsize=12)
axes[0].set_ylabel('Frequency', fontsize=12)
axes[0].set_title(f'Distribution of {TARGET.title()}', fontsize=14, fontweight='bold')
axes[0].grid(True, alpha=0.3)

# Box plot
axes[1].boxplot(df[TARGET].dropna(), vert=True)
axes[1].set_ylabel('Price ($)', fontsize=12)
axes[1].set_title(f'Box Plot of {TARGET.title()}', fontsize=14, fontweight='bold')
axes[1].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

# Check skewness and distribution
skewness = df[TARGET].skew()
kurtosis = df[TARGET].kurtosis()

print(f"\nDistribution Analysis:")
print(f"Skewness: {skewness:.3f}")
print(f"Kurtosis: {kurtosis:.3f}")

if abs(skewness) > 2:
    print("→ Target is HIGHLY skewed. Strong log transform recommended.")
elif abs(skewness) > 1:
    print("→ Target is moderately skewed. Log transform may help.")
elif abs(skewness) > 0.5:
    print("→ Target is slightly skewed.")
else:
    print("→ Target is approximately symmetric.")
```

### 3.2 Data Quality Checks

```python
# Check for duplicate rows
dups = df.duplicated().sum()
print(f"Duplicate rows: {dups:,} ({dups/len(df)*100:.3f}%)")

if dups > 0:
    print("⚠️ Duplicates found! May need to remove them during cleaning.")
else:
    print("✓ No exact duplicate rows detected.")
```

```python
# Create a summary of missing values
missing_count = df.isnull().sum()
missing_pct = (missing_count / len(df)) * 100
missing_df = pd.DataFrame({
    'Missing Count': missing_count, 
    'Missing %': missing_pct
})

# Show only columns with missing values, sorted by % missing
missing_df = missing_df[missing_df['Missing Count'] > 0].sort_values('Missing %', ascending=False)

print("Missing Values Summary (only showing columns with missing data):")
print("="*60)
display(missing_df)
```

```python
# Visualize missing values
if len(missing_df) > 0:
    plt.figure(figsize=(12, 8))
    
    # Color code by severity
    colors = []
    for pct in missing_df['Missing %']:
        if pct > 50:
            colors.append('red')
        elif pct > 20:
            colors.append('orange')
        else:
            colors.append('steelblue')
    
    plt.barh(missing_df.index, missing_df['Missing %'], color=colors, edgecolor='black')
    plt.xlabel('Missing Percentage (%)', fontsize=12)
    plt.title('Missing Values by Column', fontsize=14, fontweight='bold')
    plt.axvline(x=50, color='red', linestyle='--', alpha=0.7, linewidth=2, label='50% threshold')
    plt.axvline(x=20, color='orange', linestyle='--', alpha=0.7, linewidth=2, label='20% threshold')
    plt.legend()
    plt.grid(True, alpha=0.3, axis='x')
    plt.tight_layout()
    plt.show()
    
    # Identify problem columns
    high_missing = missing_df[missing_df['Missing %'] > 50]
    if len(high_missing) > 0:
        print(f"\n⚠️ Columns with >50% missing data (candidates for dropping):")
        print(high_missing.index.tolist())
else:
    print("✓ No missing values in the dataset!")
```
